# Importing Necessary Libraries

```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import StratifiedKFold
from lightgbm import LGBMClassifier

from scipy.sparse import hstack
from tqdm import tqdm
```

```python
train_data = pd.read_csv('/kaggle/input/competitions/smart-mcq-solver-challenge/train.csv')
test_data = pd.read_csv('/kaggle/input/competitions/smart-mcq-solver-challenge/test.csv')
submission_data = pd.read_csv('/kaggle/input/competitions/smart-mcq-solver-challenge/sample_submission.csv')
```

```python
print("Dataset preview : \n")
train_data.head()
```

```python
print(f"Train data size : {train_data.shape}")
print(f"Test data size : {test_data.shape}")
```

# Preprocessing

```python
def clean_text(text):
    text = str(text)
    text = text.lower()
    text = text.replace("\n"," ")
    text = " ".join(text.split())
    return text
```

```python
for col in ["prompt","A","B","C","D","E"]:
    train_data[col] = train_data[col].fillna("").apply(clean_text)
    test_data[col] = test_data[col].fillna("").apply(clean_text)
```

# Convert MCQ into Ranking Format

```python
options = ["A","B","C","D","E"]

train_rows = []

for _, row in tqdm(train_data.iterrows(), total=len(train_data)):
    
    correct = row["answer"]
    
    for op in options:
        
        train_rows.append({
            "id": row["id"],
            "prompt": row["prompt"],
            "option": row[op],
            "label": int(op == correct)
        })

train_long = pd.DataFrame(train_rows)

print(train_long.shape)
```

# Test Ranking Format

```python
test_rows = []

for _, row in tqdm(test_data.iterrows(), total=len(test_data)):
    
    for op in options:
        
        test_rows.append({
            "id": row["id"],
            "option_label": op,
            "prompt": row["prompt"],
            "option": row[op]
        })

test_long = pd.DataFrame(test_rows)
```

# TF-IDF Feature Engineering

```python
train_text = (
    train_long["prompt"] +
    " [SEP] " +
    train_long["option"]
)

test_text = (
    test_long["prompt"] +
    " [SEP] " +
    test_long["option"]
)
```

```python
tfidf = TfidfVectorizer(
    max_features=50000,
    ngram_range=(1,2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True,
    stop_words="english"
)
```

```python
X_train = tfidf.fit_transform(train_text)

X_test = tfidf.transform(test_text)

y = train_long["label"]
```

```python
print(type(X_train))
print(type(X_test))
```

```python
question_ids = train_long["id"]
```

# LightGBM

```python
lgb_params = {
    "n_estimators": 1000,
    "learning_rate": 0.05,
    "num_leaves": 63,
    "objective": "binary",
    "metric": "auc",
    "verbosity": -1,
    "random_state": 42
}
```

## Train

```python
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

oof = np.zeros(len(train_long))
test_preds = np.zeros(len(test_long))

for fold, (tr_idx, val_idx) in enumerate(
        skf.split(X_train, y)):
    
    print(f"Fold {fold+1}")
    
    model = LGBMClassifier(**lgb_params)
    
    model.fit(
        X_train[tr_idx],
        y.iloc[tr_idx]
    )
    
    oof[val_idx] = model.predict_proba(
        X_train[val_idx]
    )[:,1]
    
    test_preds += (
        model.predict_proba(X_test)[:,1] / 5
    )
    print(test_preds)
```

```python
test_long["score"] = test_preds
```

```python
test_long.head()
```

# Submission

```python
submission_rows = []

for qid, group in test_long.groupby("id"):
    
    ranked = group.sort_values(
        "score",
        ascending=False
    )
    
    top3 = ranked["option_label"].head(3).tolist()
    
    submission_rows.append({
        "id": qid,
        "prediction": " ".join(top3)
    })
```

```python
submission = pd.DataFrame(submission_rows)
submission.head()
```

```python
submission.to_csv(
    "submission.csv",
    index=False
)
```
